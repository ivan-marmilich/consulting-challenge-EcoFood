```python
import pandas as pd
import numpy as np
df_limpio = pd.read_csv('../data/processed/oferta_gastronomica_limpia.csv', dtype={"codigo_postal": str})

df_limpio['mail'] = df_limpio['mail'].fillna('Mail no registrado')
df_limpio['horario'] = df_limpio['horario'].fillna('Horario no registrado')
df_limpio['direccion_completa'] = df_limpio['direccion_completa'].fillna('Direccion no registrada')
df_limpio['codigo_postal'] = df_limpio['codigo_postal'].fillna('Codigo postal no registrado')
df_limpio['codigo_postal'] = 
    df_limpio['codigo_postal']
    .str.replace(r'\.0$', '', regex=True)
    .str.replace(r'\D', '', regex=True)
    .replace('', 'Codigo postal no registrado')

print(df_limpio['codigo_postal'].value_counts(dropna=False).head(6))
```
